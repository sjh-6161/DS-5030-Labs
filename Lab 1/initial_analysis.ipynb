<div class="alert alert-block alert-warning">
<b>Question 6:</b>  What is the average amount each university receives? Median amount? If they are different, why?
</div>

```python
# 1.3.6
uni_totals = gift.groupby('Institution Name')['Foreign Gift Amount'].sum()

mean_total_per_uni = uni_totals.mean()
median_total_per_uni = uni_totals.median()

print(f"Mean donation total by Uni: {mean_total_per_uni:,.2f}")
print(f"Median donation total by Uni: {median_total_per_uni:,.2f}")
```

The difference between the mean and median donations to universities is due to the distribution being skewed heavily by uncommon, large donations. This pulls the mean amount higher much faster than it does the median amount. 

**Plot a histogram of the amounts received by each university, and then label the median and the mean as vertical lines.**

```python
# Total amount per university
totals = gift.groupby('Institution Name')['Foreign Gift Amount'].sum()
mean_val = totals.mean()
median_val = totals.median()

fig = px.histogram(
    totals,
    x='Foreign Gift Amount',
    nbins=50,
    title='Distribution of Total Foreign Gift Amount Received per University'
)

# Adding vertical lines
fig.add_vline(x=mean_val, line_dash='dash', line_color='red',
              annotation_text=f'Mean: ${mean_val:,.0f}', annotation_position='top right')
fig.add_vline(x=median_val, line_dash='dash', line_color='yellow',
              annotation_text=f'Median: ${median_val:,.0f}', annotation_position='top left')

fig.show()
```

<div class="alert alert-block alert-warning">
<b>Question 7:</b>  What is the largest flow of a single country to a single institution?
</div>

```python
# Cross tab table showing relationship between countries and instutions via gifts (top 10 of each via total donations)
top10_insts = gift.groupby('Institution Name')['Foreign Gift Amount'].sum().sort_values(ascending=False).head(10)
top10_conts = gift.groupby('Country of Giftor')['Foreign Gift Amount'].sum().sort_values(ascending=False).head(10)

filtered_top10 = gift[gift['Country of Giftor'].isin(list(top10_conts.index)) & gift['Institution Name'].isin(list(top10_insts.index))]

pd.crosstab(filtered_top10['Country of Giftor'], filtered_top10['Institution Name'], values=filtered_top10['Foreign Gift Amount'], aggfunc='sum')
```

```python
# Flow chart (using example code from the Lab01.ipynb file). 

import plotly.graph_objects as go

# giftor = 'Giftor Name'
# recipi = 'Institution Name'
# flor = 'Foreign Gift Amount'

giftor = 'Giftor Name'
recipi = 'Institution Name'
flow = 'Foreign Gift Amount'
N = 20

flows = (
    gift.groupby([giftor, 
                recipi])
      [flow]
      .sum()
      .nlargest(N)
      .reset_index()
)

labels = (
    flows[giftor].tolist()
    + flows[recipi].tolist()
)

labels = list(dict.fromkeys(labels))

fig = go.Figure(
    go.Sankey(
        node=dict(label=labels),
        link=dict(
            source=flows[giftor]
                        .map(labels.index),
            target=flows[recipi]
                        .map(labels.index),
            value=flows[flow]
        )
    )
)

fig.show()
```

```python
# 1.3.7
gift.groupby(['Country of Giftor', 'Institution Name'])['Foreign Gift Amount'].sum().reset_index().sort_values('Foreign Gift Amount', ascending=False).head()
```

The largest flow of a single country to a single insitution is Qatar to Cornell. Qatar has donated $1,018,473,315 to Cornell. While the group by aggregation table provides the simplest way to get to the answer (the largest country-instituation spending), it doesn't do as good of a job at visualizing the connections. The crosstab and sankey diagram help to illustrate a fuller picture of the spending.
