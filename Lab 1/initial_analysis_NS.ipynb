### Question 6: What is the average amount each university receives? What is the median? Why might they differ?

The mean total amount received by an institution is approximately **$52.2 million**, while the median is approximately **$6.49 million**.

The mean is much larger than the median because the distribution is strongly **right-skewed**. A small number of institutions receive extremely large amounts of foreign funding, which pulls the mean upward. The median is less affected by these very large values and therefore better represents a more typical institution.

```python
university_stats = (
    df_FGE.groupby('Institution Name')['Foreign Gift Amount']
    .agg(['mean', 'median', 'count'])
    .sort_values('mean', ascending=False)
)

university_stats.head(10)
```

```python
university_stats_filtered = university_stats[
    university_stats['count'] >= 20
]

university_stats_filtered.head(10)
```

### Question 7: What is the largest flow from a single country to a single institution?

The largest flow from a single country to a single institution is from **Qatar to Cornell University**, with a total foreign gift amount of approximately **$1.02 billion**.

This result was found by grouping the data by both **Country of Giftor** and **Institution Name**, summing the foreign gift amounts for each country–institution pair, and then sorting the totals from largest to smallest.

```python
top_giftors = (
    df_FGE.groupby('Giftor Name')['Foreign Gift Amount']
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

top_giftors.head(10)
```

### Question 8: Who are the top giftors to U.S. academic institutions?

The largest named giftor in the dataset is the **Qatar Foundation**, with approximately **$1.17 billion** in total foreign gifts.

Other major giftors include **Qatar Foundation/Qatar National Research**, **Qatar Foundation for Education**, the **Saudi Arabian Cultural Mission**, and several anonymous donors.

It is important to note that some giftor names may refer to related organizations under slightly different names, so totals may be split across multiple entries.
